## El problema: reglas que cambian según el caso

Una tienda online tiene esta política: *"si la compra supera los $100.000, el envío es gratis"*


````python
compra = 120000

print("El total de la compra es", compra)
print("¿Aplica envío gratis?")
````

El `print` de la última línea es una mentira: no respondió nada, solo repitió la pregunta como texto. Hace falta una estructura que compare el número y decida.

## `if`: la pregunta con un solo camino

La traducción directa de "si" en español. `if <condición>:` seguido de un bloque indentado que *solo* corre si la condición es verdadera.

````python
compra = 120000

if compra >= 100000:
    print("Envío gratis")
````

Si cambiás `compra` a un valor menor a 100000 y volvés a correr la celda, no va a imprimir nada — porque no hay ningún otro camino definido todavía. Probalo.

````python
compra = 45000

if compra >= 100000:
    print("Envío gratis")
````

## `else`: el otro camino

Para cubrir el caso contrario sin repetir la condición al revés, se usa `else` ("si no").

````python
compra = 45000

if compra >= 100000:
    print("Envío gratis")
else:
    print("Envío con costo: $8.000")
````

Ahora sí, sea cual sea el valor de `compra`, siempre se imprime algo. Eso es lo que distingue un `if` suelto de un `if`/`else` completo: cubrís *todos* los casos, no solo el que te interesa.

## La trampa de `>` vs `>=`

La política dice "si la compra **supera** los $100.000". ¿Eso es `>` o `>=`? Probemos con el caso límite: una compra de exactamente $100.000.

````python
compra = 100000

if compra > 100000:
    print("Envío gratis")
else:
    print("Envío con costo: $8.000")

(0, 3) 3 no lo incluye (>)
[0,3] >=
    
````

Con `>` ese cliente de $100.000 exactos **no** recibe el envío gratis, aunque en la intención real de la regla probablemente sí debería. Cambiá el operador a `>=` y volvé a correr la celda: ese mismo cliente cambia de resultado sin que cambie ni un peso de su compra. Un operador mal elegido no es un detalle de sintaxis: cambia una decisión de negocio real.

## `elif`: más de dos caminos

Las reglas reales casi nunca son binarias. Agreguemos un tercer nivel: compras sobre $300.000 tienen, además, envío prioritario.

````python
compra = 350000

if compra >= 300000:
    print("Envío gratis + prioritario")
elif compra >= 100000:
    print("Envío gratis")
else:
    print("Envío con costo: $8.000")
````

El orden importa. Python evalúa de arriba a abajo y se queda con el **primer** bloque que sea verdadero — ignora los demás aunque también lo sean. Por eso la condición más exigente (`>= 300000`) va primero: si pusieras `>= 100000` arriba, una compra de $350.000 nunca llegaría a ver el `elif` de prioritario, porque ya habría entrado al primer bloque que encontró verdadero.

````python
# mismo caso, pero con el orden invertido — fijate qué responde
compra = 350000

if compra >= 100000:
    print("Envío gratis")
elif compra >= 300000:
    print("Envío gratis + prioritario")
else:
    print("Envío con costo: $8.000")
````

## `and` / `or`: combinar condiciones

Otra regla típica: envío gratis y prioritario solo si la compra es grande **y** el cliente es de una ciudad principal.

````python
compra = 350000
ciudad = "Barranquilla"
ciudades_principales = ["Bogotá", "Medellín", "Cali"]

if compra >= 300000 and ciudad in ciudades_principales:
    print("Envío gratis + prioritario")
else:
    print("Envío estándar")
````

`and` exige que **ambas** condiciones sean verdaderas. `or` exige que **al menos una** lo sea. Cambiá `ciudad` a una que no esté en la lista y volvé a correr: con `and` el resultado cambia; probá después cambiar `and` por `or` y ver cómo cambia otra vez, sin tocar ningún otro valor.

## Lo que se rompe

Antes de seguir a los loops, los errores clásicos de `if`/`else` que vas a cometer tarde o temprano.

Primero, el más común: confundir `=` (asignar) con `==` (comparar).

````python
compra = 100000

if compra = 100000:
    print("Envío gratis")

    
````

`SyntaxError: invalid syntax`. `=` guarda un valor en una variable; `==` pregunta si dos valores son iguales. Son dos operaciones completamente distintas que se parecen demasiado al escribirlas rápido.

````python
compra = 100000

if compra == 100000:
    print("Envío gratis")
````

Segundo: olvidar los dos puntos al final de la línea del `if`.

````python
compra = 100000

if compra >= 100000
    print("Envío gratis")
````

`SyntaxError: expected \':\'`. Python necesita los dos puntos para saber dónde termina la condición y empieza el bloque que depende de ella.

Tercero: la indentación no es decorativa, es la forma en que Python sabe qué línea pertenece a qué bloque.

````python
compra = 4500

if compra >= 100000:
    print("Envío gratis")
print("Gracias por tu compra")
````

Esto corre sin error, pero fijate que "Gracias por tu compra" se imprime *siempre*, incluso con una compra chica — porque al no estar indentada, esa línea ya no pertenece al `if`, es una instrucción aparte que corre siempre. La indentación no es estética: define qué código depende de la condición y qué código no.

## `for`: repetir una acción por cada elemento

Ahora el otro problema del día: revisar el stock de varios productos, uno por uno, sin copiar y pegar la misma línea 500 veces.

````python
inventario = {"mouse": 15, "teclado": 3, "monitor": 0, "audífonos": 8}

for producto in inventario:
    print(producto)
````

`for producto in inventario:` toma cada elemento de la colección, uno a la vez, y lo guarda temporalmente en `producto` mientras corre el bloque indentado. Cuando se acaban los elementos, el loop termina solo.

## Combinando `for` con `if`: la razón por la que importan juntos

La regla real no es "mostrar el inventario", es "avisar cuáles productos están bajos de stock". Eso es un `for` que, en cada vuelta, le hace una pregunta a un `if`.

````python
inventario = {"mouse": 15, "teclado": 3, "monitor": 0, "audífonos": 8}

for producto in inventario:
    stock = inventario[producto]
    if stock < 10:
        print("⚠️ Avisar: stock bajo de", producto, "—", stock, "unidades")
````

Esa es la combinación que de verdad se usa en el mundo real: el `for` recorre, el `if` decide. Ningún humano revisó los 4 productos a mano — y en un inventario de 5.000 productos, el código es exactamente el mismo.

## `while`: repetir mientras algo sea verdad

`for` repite un número conocido de veces (una por elemento). `while` repite *mientras* una condición siga siendo verdadera, sin saber de antemano cuántas vueltas va a dar.

````python
stock = 5

while stock > 0:
    print("Queda stock:", stock)
    stock = stock - 1

print("Se agotó el stock")
````

## El loop infinito: el error más peligroso de esta sesión

¿Qué pasa si te olvidás de la línea que cambia `stock`? La condición nunca deja de ser verdadera, y el loop nunca termina. **No corras la siguiente celda** — léela nomás. Si la corrieras, tendrías que interrumpir la ejecución manualmente (el botón de stop del notebook).

```python
stock = 5

while stock > 0:
    print("Queda stock:", stock)
    # falta: stock = stock - 1
```

Sin esa línea, `stock` siempre vale 5, la condición `stock > 0` siempre es verdadera, y el `print` corre para siempre. Todo `while` necesita algo dentro del bloque que eventualmente lo haga falso.

## `break` y `continue`: salidas de emergencia dentro de un loop

`break` corta el loop por completo, de inmediato. `continue` salta solo la vuelta actual y sigue con la siguiente.

````python
inventario = {"mouse": 15, "teclado": 3, "monitor": 0, "audífonos": 8}

# break: paramos en cuanto encontramos el primer producto agotado
for producto in inventario:
    stock = inventario[producto]
    if stock == 0:
        print("Agotado:", producto, "— deteniendo revisión")
        break
    print("Revisado:", producto)
````

````python
inventario = {"mouse": 15, "teclado": 3, "monitor": 0, "audífonos": 8}

# continue: saltamos los agotados, pero seguimos revisando el resto
for producto in inventario:
    stock = inventario[producto]
    if stock == 0:
        continue
    print("Revisado:", producto, "—", stock, "unidades")
````
