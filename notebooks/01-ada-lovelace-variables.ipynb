## ¿Para qué una variable, si podés escribir el texto directo?

Mirá este código. No usa ninguna variable, el nombre de Ada está escrito tres veces, directo:

```python
print("Ada Lovelace escribió el primer algoritmo.")
print("Ada Lovelace nació en 1815.")
print("El trabajo de Ada Lovelace sigue citado hoy.")
```

Funciona. Pero ahora imaginate que te diste cuenta de que escribiste mal el nombre, o que en realidad querés usar "Ada Byron" en todo el texto. Tenés que ir línea por línea, a mano, cambiándolo en cada lugar donde aparece. Acá son 3 líneas, fácil. En un programa de verdad pueden ser cientos.

Con una variable, el nombre vive en un solo lugar:

```python
nombre = "Ada Lovelace"

print(nombre, "escribió el primer algoritmo.")
print(nombre, "nació en 1815.")
print("El trabajo de", nombre, "sigue citado hoy.")
```

## Guardar un dato en una variable



```python
nombre = "Ada Lovelace"

print(nombre)
```

Esa "caja" existe de verdad. No es solo una forma de explicarlo.

Cuando corriste `nombre = "Ada Lovelace"`, Python guardó el texto en un espacio de la RAM, la memoria de trabajo del computador (no la ROM: esa es otra cosa, guarda el firmware y no se toca al programar). Ese espacio tiene una dirección, como si fuera un número de apartamento. `nombre` no es el texto, es la etiqueta que apunta a esa dirección.

Las direcciones de memoria se escriben en hexadecimal, algo como `0x7f8a3c061b70`:

```python
from IPython.display import Image
Image(filename="assets/ram-diagram.png", width=1000)
```

Podés ver la dirección real de tu variable con `id()`, convertida a hex:

```python
nombre = "Ada Lovelace"

print(hex(id(nombre)))
```

Eso que empieza con `0x` es la dirección en RAM donde vive el texto ahora mismo. Si cerrás el notebook y lo volvés a abrir, va a ser otra dirección: la RAM se vacía al apagar, por eso no es donde se guardan los archivos de verdad.

Cuando más adelante veas un `NameError` porque usaste una variable que no creaste, es literal: no hay dirección reservada bajo ese nombre, Python no tiene dónde ir a buscar.

Una trampa con esto: a veces dos variables distintas comparten dirección, y a veces no, y no es al azar.

```python
a = 5
b = 5
print(a is b)

texto1 = "este es un texto bastante largo para probar"
texto2 = "este es un texto bastante largo para probar"
print(texto1 is texto2)
```

`a is b` da `True`: Python reutiliza la misma dirección para números chicos, porque los usa tanto que los deja precargados en memoria. `texto1 is texto2` normalmente da `False`, aunque el contenido sea idéntico: cada texto se guardó en su propia dirección.

No cambia en nada cómo programás (para comparar valores siempre usás `==`, no `is`), pero explica por qué `id()` a veces "miente" y parece que dos variables son la misma caja sin serlo.

## Varias variables a la vez

Sigamos guardando cosas sobre ella, un dato por variable.

```python
nombre = "Ada Lovelace"
nacio_en = 1815
pais = "Inglaterra"
escribio_en = 1843

print(nombre)
print(nacio_en)
print(pais)
print(escribio_en)
```

Ada trabajó con Charles Babbage en la Máquina Analítica, un diseño que nunca se terminó de construir en vida de ninguno de los dos. En las notas que ella escribió sobre esa máquina está lo que hoy se considera el primer algoritmo pensado para una máquina. No alcanzó a verla funcionar.

## Combinar las variables en un mensaje

Tener los datos sueltos está bien, pero contar algo requiere juntarlos.

```python
nombre = "Ada Lovelace"
nacio_en = 1815
pais = "Inglaterra"
escribio_en = 1843
que_escribio = "el primer algoritmo pensado para una máquina"

print("Nombre:", nombre)
print("Nació en", nacio_en, "en", pais)
print("En", escribio_en, "escribió", que_escribio)
print("Eso la convierte en la primera programadora de la historia.")
```

Fijate que es el mismo orden que seguramente seguirías si le contaras esto a una amiga: quién es, cuándo nació, qué hizo. Programar no inventa una lógica nueva, usa la que ya tenés.

## Lo que se rompe (y por qué)

Antes de seguir, veamos cosas que te van a pasar tarde o temprano con variables. Mejor ahora, con Ada, que sola en tu casa sin saber qué significa el error.

Primero, un clásico: poner el nombre de la variable entre comillas.

```python
nombre = "Ada Lovelace"

print("nombre")
```

Eso imprimió la palabra "nombre", no "Ada Lovelace". Las comillas le dicen a Python que es texto fijo, no que vaya a buscar la variable. Sin comillas alrededor del nombre de la variable:

```python
print(nombre)
```

Segundo: usar una variable que todavía no existe.

```python
print(apellido)
```

`NameError: name 'apellido' is not defined`. Es el mismo tema de la RAM del principio: `apellido` nunca tuvo una dirección reservada, porque nunca corriste un `=` con ese nombre. Python no busca en todo el computador a ver si adivina qué quisiste decir, solo mira las direcciones que ya existen. El código corre de arriba hacia abajo, en ese orden, así que si usás una variable antes de crearla, tampoco existe todavía.

```python
apellido = "Lovelace"
print(apellido)
```

Tercero, mezclar texto y número con `+`:

```python
nacio_en = 1815

print("Ada nació en " + nacio_en)
```

`TypeError: can only concatenate str (not "int") to str`. Python no suma texto con número usando `+`. Por eso en todo este notebook veníamos separando por comas dentro del `print()`, en vez de pegar todo junto.

```python
print("Ada nació en", nacio_en)
```

Cuarto, el más traicionero porque no tira ningún error: reasignar una variable sin darte cuenta.

```python
pais = "Inglaterra"
print(pais)

pais = "Reino Unido"
print(pais)
```

La segunda línea pisó la primera. `"Inglaterra"` se perdió para siempre, no hay forma de recuperarlo desde `pais`, porque la etiqueta ahora apunta a otra dirección en RAM. Python no te avisa, no hay ningún error: simplemente dejaste de poder ver el valor viejo. Si en un notebook largo reusás el mismo nombre de variable en dos partes distintas sin querer, es la causa más común de "pero yo esto ya lo tenía calculado" cuando no aparece.

Quinto: un nombre de variable que Python no acepta.

```python
1nombre = "Ada Lovelace"
```

`SyntaxError: invalid decimal literal`. En Python un nombre de variable no puede empezar con número, ni tener espacios, ni usar guiones medios. `1nombre` no sirve, `mi nombre` tampoco, `mi-nombre` tampoco. La convención es `nombre_apellido`, con guión bajo en vez de espacio.

Sexto: para Python, mayúscula y minúscula son letras distintas, no la misma letra "más grande".

```python
Ciudad = "Londres"

print(ciudad)
```

`NameError: name 'ciudad' is not defined`, aunque arriba sí creaste `Ciudad`, con mayúscula. Para Python son dos cajas distintas, dos direcciones distintas. Si venís de Excel, donde `NOMBRE` y `nombre` son lo mismo, esto sorprende la primera vez. Acá no: cada combinación exacta de mayúsculas y minúsculas es una variable separada.

Si te pasó alguno de estos hoy en casa, no es que lo hiciste mal, es literalmente la clase de hoy. Mandá el error completo al grupo si te traba, seguro ya le pasó a alguien más.
